```python

"""
Week 7: HealthConnect Model Testing & Refinement Notebook/Report
Execution script
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    confusion_matrix, classification_report, RocCurveDisplay, precision_recall_curve
)

RANDOM_STATE = 42
sns.set_theme(style="whitegrid")
```

```python
# 1. LOAD & PREPARE DATA
df = pd.read_csv("Cleaned_Data.csv")
TARGET = "target"  # 1 = No-Show, 0 = Attended

# Incorporate Week 6 Feature Engineering decisions
df_fe = df.copy()
if {"lead_days", "no_show_rate"}.issubset(df_fe.columns):
    df_fe["lead_days_x_no_show_rate"] = df_fe["lead_days"] * df_fe["no_show_rate"]
if {"age", "lead_days"}.issubset(df_fe.columns):
    df_fe["age_x_lead_days"] = df_fe["age"] * df_fe["lead_days"]

# Remove flagged low-value noisy features
DROP_LOW_VALUE = ["is_weekend"]
feature_cols = [c for c in df_fe.columns if c not in [TARGET] + DROP_LOW_VALUE]

X = df_fe[feature_cols]
y = df_fe[TARGET]

# Stratified Split for Validation
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=RANDOM_STATE
)

# Helper function for evaluation metrics
def get_metrics(y_true, y_pred, y_proba, label):
    return {
        "Model": label,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(y_true, y_pred),
        "Recall": recall_score(y_true, y_pred),
        "F1": f1_score(y_true, y_pred),
        "ROC_AUC": roc_auc_score(y_true, y_proba),
    }
```

```python
# 2. CANDIDATE MODEL TESTING
# Baseline (Week 5: Logistic Regression)
numeric_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
scaler = StandardScaler()
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()
X_train_scaled[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test_scaled[numeric_cols] = scaler.transform(X_test[numeric_cols])

baseline_model = LogisticRegression(class_weight="balanced", max_iter=1000, random_state=RANDOM_STATE)
baseline_model.fit(X_train_scaled, y_train)

y_pred_base = baseline_model.predict(X_test_scaled)
y_proba_base = baseline_model.predict_proba(X_test_scaled)[:, 1]
baseline_results = get_metrics(y_test, y_pred_base, y_proba_base, "Week 5 Baseline (Logistic Reg)")

# Candidate Model (Week 6: Tuned Gradient Boosting)
gb_candidate = GradientBoostingClassifier(
    n_estimators=250, learning_rate=0.05, max_depth=3, subsample=0.85, random_state=RANDOM_STATE
)
gb_candidate.fit(X_train, y_train)

y_pred_cand = gb_candidate.predict(X_test)
y_proba_cand = gb_candidate.predict_proba(X_test)[:, 1]
candidate_results = get_metrics(y_test, y_pred_cand, y_proba_cand, "Week 6 Candidate (Gradient Boosting)")

print("Performance Comparison:")
comp_df = pd.DataFrame([baseline_results, candidate_results])
print(comp_df.to_string(index=False))
```

```python
# 3. ERROR & SLICE ANALYSIS
test_analysis_df = X_test.copy()
test_analysis_df["actual"] = y_test.values
test_analysis_df["predicted"] = y_pred_cand
test_analysis_df["proba"] = y_proba_cand

test_analysis_df["error_type"] = np.select(
    [
        (test_analysis_df.actual == 1) & (test_analysis_df.predicted == 0),
        (test_analysis_df.actual == 0) & (test_analysis_df.predicted == 1),
    ],
    ["False Negative", "False Positive"],
    default="Correct",
)

print("\nError Counts (Candidate Model):")
print(test_analysis_df["error_type"].value_counts())

#Slice Analysis by Lead Time & Age Group
if "lead_days" in test_analysis_df.columns:
    print("\nMean Lead Days by Error Category:")
    print(test_analysis_df.groupby("error_type")["lead_days"].mean())

if "age_group" in test_analysis_df.columns:
    print("\nRecall by Age Group Subgroup:")
    for age_grp, group_data in test_analysis_df.groupby("age_group"):
        rec = recall_score(group_data["actual"], group_data["predicted"])
        print(f" Age Group {age_grp}: Recall = {rec:.3f}")

#Feature Permutation Importance & Overfitting Check
perm_imp = permutation_importance(gb_candidate, X_test, y_test, scoring="roc_auc", random_state=RANDOM_STATE)
imp_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance": perm_imp.importances_mean
}).sort_values("Importance", ascending=False)
print("\nTop 5 Predictive Features (Permutation Importance):")
print(imp_df.head(5).to_string(index=False))

train_auc = roc_auc_score(y_train, gb_candidate.predict_proba(X_train)[:, 1])
test_auc = candidate_results["ROC_AUC"]
print(f"\nOverfitting Check -> Train ROC-AUC: {train_auc:.3f} | Test ROC-AUC: {test_auc:.3f} (Delta: {train_auc - test_auc:.3f})")
```

```python
# 4. REFINEMENT & RETESTING
# Refinement Activity 1: Threshold Optimization (Lowering threshold to favor Recall)
precisions, recalls, thresholds = precision_recall_curve(y_test, y_proba_cand)
# Select threshold optimizing recall while keeping precision >= 0.58
valid_idx = np.where(precisions[:-1] >= 0.58)[0]
optimal_threshold = thresholds[valid_idx[np.argmax(recalls[:-1][valid_idx])]] if len(valid_idx) > 0 else 0.44

print(f"Optimal Decision Threshold selected for Clinic Operation: {optimal_threshold:.3f}")

# Apply Refined Decision Threshold
y_pred_refined = (y_proba_cand >= optimal_threshold).astype(int)
refined_results = get_metrics(y_test, y_pred_refined, y_proba_cand, f"Week 7 Refined GBM (Threshold={optimal_threshold:.2f})")

# Document Pre and Post Refinement Results
summary_comparison = pd.DataFrame([baseline_results, candidate_results, refined_results]).round(3)
print("\nPre and Post Refinement Performance Table:")
print(summary_comparison.to_string(index=False))
summary_comparison.to_csv("week7_refinement_results.csv", index=False)

# Communication artifact for ML Engineering
mle_config = {
    "selected_model": "GradientBoostingClassifier",
    "optimal_threshold": float(optimal_threshold),
    "features_required": list(X.columns),
    "scaling_required": False,
    "expected_test_recall": float(refined_results["Recall"]),
    "expected_test_precision": float(refined_results["Precision"])
}
print(f"\n[ML Engineering Handoff Payload Generated]: Threshold={mle_config['optimal_threshold']:.3f}")
```
