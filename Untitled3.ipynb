```python
import pandas as pd
import numpy as np

# 1. Load the dataset
df = pd.read_csv('HealthData.csv')

print(f"Initial shape: {df.shape}")

#STEP 1: ADDRESS DATA LEAKAGE & IRRELEVANT VARIABLES
# Based on the design doc: we predict AT THE TIME OF BOOKING.
# 'waiting_time_minutes' is only known after they arrive.
# 'reminder_sent' and 'reminder_channel' happen after booking.
# IDs are not predictive.
leakage_and_irrelevant = [
    'appointment_id',
    'patient_id',
    'waiting_time_minutes',
    'reminder_sent',
    'reminder_channel'
]
df.drop(columns=leakage_and_irrelevant, inplace=True)


#STEP 2: TREATMENT OF CANCELLATIONS
# Per Week 5 analysis: remove 'Cancelled' to focus strictly on No-Show vs. Attended
df = df[df['appointment_outcome'] != 'Cancelled'].copy()

# Convert target to binary: No-Show = 1, Attended = 0
df['target'] = df['appointment_outcome'].map({'No-Show': 1, 'Attended': 0})
df.drop(columns=['appointment_outcome'], inplace=True)


#STEP 3: MISSING VALUES
# distance_to_clinic_km has missing values. We will use the median.
median_dist = df['distance_to_clinic_km'].median()
df['distance_to_clinic_km'] = df['distance_to_clinic_km'].fillna(median_dist)


#STEP 4: INCORRECT DATA TYPES & INCONSISTENT RECORDS
# Convert dates to datetime objects
df['booking_date'] = pd.to_datetime(df['booking_date'])
df['appointment_date'] = pd.to_datetime(df['appointment_date'])

# Sanity Check: Ensure appointment isn't before booking
# If lead_days < 0, it's an inconsistent record
df = df[df['booking_lead_days'] >= 0]

# Address "Prefer not to say" in Gender by grouping with a new category or mode
df['gender'] = df['gender'].replace('Prefer not to say', 'Other')


#STEP 5: CATEGORICAL VARIABLES (ENCODING)
# Binary Encoding for Gender
df['is_female'] = df['gender'].map({'Female': 1, 'Male': 0, 'Other': 0.5})

# One-Hot Encoding for multi-class categories
df = pd.get_dummies(df, columns=['appointment_type', 'age_group'], drop_first=True)

# Ordinal Encoding for Appointment Time (Morning < Afternoon < Evening)
time_map = {'Morning': 1, 'Afternoon': 2, 'Evening': 3}
df['appointment_time_score'] = df['appointment_time'].map(time_map)

# Cyclical Encoding for Day of Week (Monday=0, Sunday=6)
# First convert to a number
day_map = {
    'Monday': 0, 'Tuesday': 1, 'Wednesday': 2,
    'Thursday': 3, 'Friday': 4, 'Saturday': 5, 'Sunday': 6
}
df['day_num'] = df['appointment_day'].map(day_map)


#STEP 6: FINAL CLEANUP
# Remove the original string/date columns that have been encoded
final_df = df.drop(columns=[
    'gender', 'appointment_day', 'appointment_time',
    'booking_date', 'appointment_date'
])

# Remove any remaining duplicates
final_df.drop_duplicates(inplace=True)

print(f"Final shape for modeling: {final_df.shape}")
print("\nFirst 5 rows of prepared data:")
print(final_df.head())

# Export for modeling
final_df.to_csv('Cleaned_Data.csv', index=False)
```

Train/Test Strategy:
Stratified Random Split(80/20)
We split the dataset into 80% for training the model and 20% for testing it's performance on unseen data.We use stratification on the target variable.
This is appropriate because:
1.Class Imbalance:No-Shows are less frequent than attended outcomes.Stratification ensures that the training set and test set both contain the same percentage of No-Shows as the original data.
2.Generalization:an 80/20 split provides enough data for the model to learn patterns while reserving a significant portion to validately measure how well the model works in the real world.
